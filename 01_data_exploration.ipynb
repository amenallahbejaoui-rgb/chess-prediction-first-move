```python
import polars as pl

df_sample = pl.read_csv(
    "file.csv",
    n_rows=5,
    infer_schema_length=0
)

print("Shape:", df_sample.shape)
print("\nColumns:")
print(df_sample.columns)

print("\nFirst 5 games:")
display(df_sample)
```

```python
print("Dataset columns:", len(df_sample.columns))

print("\nMissing values in the sample:")
print(df_sample.null_count())

print("\nFirst Event value:")
print(df_sample["Event"][0])

print("\nFirst Site value:")
print(df_sample["Site"][0])

print("\nFirst result values:")
print(df_sample["Result"].to_list())
```

```python
valid_results = ["1-0", "0-1", "1/2-1/2", "*"]

print("Result distribution in sample:")
print(df_sample.group_by("Result").len())

print("\nValid result labels:")
print(
    df_sample.select(
        pl.col("Result")
        .is_in(valid_results)
        .all()
        .alias("All results valid")
    )
)

print("\nFirst game's moves:")
print(df_sample["movetext"][0][:500])
```

```python
# Convert ratings to numbers for analysis
df_check = df_sample.with_columns(
    pl.col("WhiteElo").cast(pl.Int64, strict=False),
    pl.col("BlackElo").cast(pl.Int64, strict=False),
)

# Create a simple outcome label
df_check = df_check.with_columns(
    pl.when(pl.col("Result") == "1-0")
    .then(pl.lit("White wins"))
    .when(pl.col("Result") == "0-1")
    .then(pl.lit("Black wins"))
    .when(pl.col("Result") == "1/2-1/2")
    .then(pl.lit("Draw"))
    .otherwise(pl.lit("Unknown"))
    .alias("Outcome")
)

display(
    df_check.select([
        "WhiteElo",
        "BlackElo",
        "ECO",
        "Opening",
        "Result",
        "Outcome",
    ])
)
```

```python
df_check = df_check.with_columns(
    (pl.col("WhiteElo") - pl.col("BlackElo"))
    .alias("RatingDifference")
)

display(
    df_check.select([
        "WhiteElo",
        "BlackElo",
        "RatingDifference",
        "Opening",
        "Outcome",
    ])
)
```

```python
sample = pl.read_csv(
    "file.csv",
    n_rows=10_000,
    infer_schema_length=0
)

print(f"Rows loaded: {sample.height:,}")
print(f"Columns: {sample.width}")
print(f"Memory used by sample: {sample.estimated_size('mb'):.2f} MB")
```

```python
print("Missing values by column:")
display(sample.null_count())

print("\nGame results:")
display(
    sample.group_by("Result")
    .len()
    .sort("len", descending=True)
)

print("\nPlayer rating summary:")
display(
    sample.select([
        pl.col("WhiteElo").cast(pl.Int64, strict=False).min().alias("Min White Elo"),
        pl.col("WhiteElo").cast(pl.Int64, strict=False).median().alias("Median White Elo"),
        pl.col("WhiteElo").cast(pl.Int64, strict=False).max().alias("Max White Elo"),
        pl.col("BlackElo").cast(pl.Int64, strict=False).min().alias("Min Black Elo"),
        pl.col("BlackElo").cast(pl.Int64, strict=False).median().alias("Median Black Elo"),
        pl.col("BlackElo").cast(pl.Int64, strict=False).max().alias("Max Black Elo"),
    ])
)
```
